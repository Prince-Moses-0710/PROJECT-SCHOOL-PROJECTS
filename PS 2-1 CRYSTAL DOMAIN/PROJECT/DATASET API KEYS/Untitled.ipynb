```python
import os

# Your extracted repository location
ELEMENT_DATA_DIR = r"C:\Users\ADMIN\Downloads\machine-learning-the-voltage-of-electrode-materials-master\dataset\element_data"

BCC_PROPERTIES = [
    "BCCenergydiff",
    "BCCefflatcnt",
    "BCCfermi",
    "BCCmagmom",
    "BCCvolume_pa"
]

# Required working ions
REQUIRED_IONS = ["Li", "Na", "K", "Ca", "Mg", "Al", "Y", "Zn"]

# Atomic numbers
atomic_numbers = {
    "Li": 3,
    "Na": 11,
    "K": 19,
    "Ca": 20,
    "Mg": 12,
    "Al": 13,
    "Y": 39,
    "Zn": 30
}

def load_element_table(property_name):
    path = os.path.join(ELEMENT_DATA_DIR, property_name + ".table")

    values = []

    with open(path, "r") as f:
        for line in f:
            line = line.strip()

            if line == "" or line == "Missing":
                values.append(0.0)
            else:
                values.append(float(line))

    return values


# Load all five BCC tables
bcc_data = {}

for prop in BCC_PROPERTIES:
    bcc_data[prop] = load_element_table(prop)

print("BCC tables loaded successfully!\n")

# Check the requested working ions
for ion in REQUIRED_IONS:
    index = atomic_numbers[ion] - 1

    print(f"\n{ion} (Atomic Number {atomic_numbers[ion]})")

    for prop in BCC_PROPERTIES:
        print(f"  {prop}: {bcc_data[prop][index]}")
```

```python
import numpy as np
from pymatgen.core import Element


def weighted_element_property(composition, property_name):
    """
    Calculate composition-weighted mean of a Magpie elemental property.
    """

    amounts = composition.get_el_amt_dict()

    values = []
    weights = []

    for symbol, amount in amounts.items():

        try:
            value = magpie.get_elemental_property(
                Element(symbol),
                property_name
            )

            # Skip unavailable values
            if value is None or not np.isfinite(value):
                continue

            values.append(float(value))
            weights.append(float(amount))

        except Exception:
            continue

    if len(values) == 0:
        return np.nan

    return np.average(values, weights=weights)
```

```python
# ============================================================
# COMPLETE 42-FEATURE CALCULATOR
# ============================================================

def calculate_42_features(composition):
    
    result = {}

    # --------------------------------------------------------
    # 37 MAGPIE FEATURES
    # --------------------------------------------------------
    for feature_name, magpie_name in FEATURE_MAP.items():

        # BCC properties are handled separately below
        if magpie_name in BCC_PROPERTIES:
            continue

        try:
            value = weighted_element_property(
                composition,
                magpie_name
            )

            result[feature_name] = value

        except Exception:
            result[feature_name] = np.nan

    # --------------------------------------------------------
    # 5 GITHUB BCC FEATURES
    # --------------------------------------------------------
    amounts = composition.get_el_amt_dict()

    for feature_name, bcc_property in [
        ("BCC energy difference", "BCCenergydiff"),
        ("BCC effective lattice constant", "BCCefflatcnt"),
        ("BCC Fermi", "BCCfermi"),
        ("BCC magnetic moment", "BCCmagmom"),
        ("BCC volume per atom", "BCCvolume_pa")
    ]:

        values = []
        weights = []

        for symbol, amount in amounts.items():

            try:
                value = bcc_lookup[bcc_property][symbol]

                if value is not None and np.isfinite(value):
                    values.append(float(value))
                    weights.append(float(amount))

            except Exception:
                continue

        if values:
            result[feature_name] = np.average(
                values,
                weights=weights
            )
        else:
            result[feature_name] = np.nan

    return result
```

```python
def weighted_element_property(composition, property_name):
    """
    Calculate composition-weighted mean of a Magpie elemental property.
    Handles list-valued properties such as OxidationStates.
    """

    amounts = composition.get_el_amt_dict()

    values = []
    weights = []

    for symbol, amount in amounts.items():

        try:
            value = magpie_data.get_elemental_property(
                Element(symbol),
                property_name
            )

            # Handle list-valued properties
            if isinstance(value, (list, tuple, np.ndarray)):
                if len(value) == 0:
                    continue
                value = value[0]

            if value is not None and np.isfinite(value):
                values.append(float(value))
                weights.append(float(amount))

        except Exception as e:
            print(
                "Problem with",
                symbol,
                property_name,
                ":",
                e
            )

    if not values:
        return np.nan

    return np.average(values, weights=weights)
```

```python
def get_weighted_feature(formula, feature_name):
    """
    Calculate the composition-weighted mean of a repository
    elemental property.
    """

    if formula is None or pd.isna(formula):
        return np.nan

    try:
        composition = Composition(str(formula))
    except Exception:
        return np.nan

    amounts = composition.get_el_amt_dict()
    total_atoms = sum(amounts.values())

    if total_atoms == 0:
        return np.nan

    weighted_sum = 0.0
    valid_weight = 0.0

    for symbol, amount in amounts.items():

        try:
            element = Element(symbol)
            Z = element.Z

            value = repository_data[feature_name][Z - 1]

            if pd.isna(value):
                continue

            fraction = amount / total_atoms

            weighted_sum += value * fraction
            valid_weight += fraction

        except Exception:
            continue

    if valid_weight == 0:
        return np.nan

    return weighted_sum / valid_weight
```

```python
def get_weighted_feature(formula, feature_name):
    """
    Calculate the feature using the original repository logic.

    OxidationStates is explicitly treated as 0 in the
    original reproduction code.
    """

    if formula is None or pd.isna(formula):
        return np.nan

    try:
        composition = Composition(str(formula))
    except Exception:
        return np.nan

    # Original repository behavior
    if feature_name == "Oxidation states":
        return 0.0

    amounts = composition.get_el_amt_dict()
    total_atoms = sum(amounts.values())

    if total_atoms == 0:
        return np.nan

    weighted_sum = 0.0

    for symbol, amount in amounts.items():

        try:
            element = Element(symbol)
            Z = element.Z

            value = repository_data[feature_name][Z - 1]

            if pd.isna(value):
                return np.nan

            weighted_sum += value * amount

        except Exception:
            return np.nan

    return weighted_sum / total_atoms
```

```python
def calculate_42_features_for_row(row):
    """
    Calculate the 42 selected features for one electrode record.

    41 features come from the original repository element tables.
    Space Group Number comes from the actual MP crystal structure.
    """

    formula = row["Framework_Formula"]
    structure = row["Crystal_Structure"]

    feature_values = {}

    # 41 repository-based features
    for feature_name in FEATURE_TABLES:
        feature_values[feature_name] = get_weighted_feature(
            formula,
            feature_name
        )

    # Feature #10: actual crystal structure space group
    try:
        feature_values["Space group number"] = (
            structure.get_space_group_info()[1]
        )
    except Exception:
        feature_values["Space group number"] = np.nan

    # Exact feature order
    return {
        feature: feature_values[feature]
        for feature in FEATURE_ORDER
    }
```

```python
# Required working ions
REQUIRED_IONS = ["Li", "Na", "K", "Ca", "Mg", "Al", "Y", "Zn"]

# Convert all_electrodes to a dataframe with their original positions
all_records = []

for idx, e in enumerate(all_electrodes):
    all_records.append({
        "original_index": idx,
        "Working_Ion": str(e.working_ion),
        "Material_ID_Charge": str(e.id_charge),
        "Material_ID_Discharge": str(e.id_discharge)
    })

all_records_df = pd.DataFrame(all_records)

# Keep only the 8 required ions
all_records_df = all_records_df[
    all_records_df["Working_Ion"].isin(REQUIRED_IONS)
].copy()

print("Available records:", len(all_records_df))

print("\nAvailable by ion:")
print(
    all_records_df["Working_Ion"]
    .value_counts()
    .reindex(REQUIRED_IONS)
    .fillna(0)
    .astype(int)
)
```

```python
# Calculate proportional target counts
available_counts = (
    all_records_df["Working_Ion"]
    .value_counts()
    .reindex(REQUIRED_IONS)
)

target_counts = (
    available_counts / available_counts.sum() * 3977
).round().astype(int)

# Correct rounding so total is exactly 3977
difference = 3977 - target_counts.sum()

if difference != 0:
    # Add/remove the difference from the largest available group
    largest_ion = available_counts.idxmax()
    target_counts[largest_ion] += difference

print("Target counts:")
print(target_counts)

print("\nTarget total:", target_counts.sum())
```

```python
selected_indices = []

for ion in REQUIRED_IONS:

    ion_records = all_records_df[
        all_records_df["Working_Ion"] == ion
    ]

    n = int(target_counts[ion])

    # Deterministic selection — no random records
    selected_indices.extend(
        ion_records.iloc[:n]["original_index"].tolist()
    )

print("Selected records:", len(selected_indices))
```

```python
selected_electrodes = [
    all_electrodes[i]
    for i in selected_indices
]

records = []

for e in selected_electrodes:
    records.append({
        "Material_ID_Charge": str(e.id_charge),
        "Material_ID_Discharge": str(e.id_discharge),
        "Working_Ion": str(e.working_ion),
        "Electrode_Formula": e.battery_formula,
        "Framework_Formula": e.framework_formula,
        "Voltage_V": e.average_voltage,
        "Frac_Charge": e.fracA_charge,
        "Frac_Discharge": e.fracA_discharge,
        "Ion_Concentration": abs(
            e.fracA_discharge - e.fracA_charge
        ),
        "Crystal_Structure": e.host_structure
    })

base_df = pd.DataFrame(records)

print("New base rows:", len(base_df))

print("\nWorking ions:")
print(base_df["Working_Ion"].value_counts())
```

```python
# Generate the 42 selected features for the NEW 3977-row base_df

feature_rows = []

for i, row in base_df.iterrows():
    feature_rows.append(calculate_42_features_for_row(row))

features_df = pd.DataFrame(feature_rows, index=base_df.index)

print("Feature dataframe shape:", features_df.shape)
print("Expected:", (3977, 42))
```
