```python
# ============================================
# FINAL ML PREPROCESSING
# ============================================

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# Load original dataset
csv_path = r"C:\Users\ADMIN\Desktop\PS PROJECT\PS 2-1 CRYSTAL DOMAIN\PROJECT\DATASET API KEYS\battery_voltage_dataset_3977.csv"

df = pd.read_csv(csv_path)

# 42 physical features
feature_cols = [
    "Atomic volume",
    "Atomic number",
    "Atomic weight",
    "Covalent radius",
    "Oxidation states",
    "Boiling temperature",
    "Column number",
    "Row number",
    "First ionization energy",
    "Space group number",
    "Density",
    "Mendeleev number",
    "ICSD volume",
    "Polarisability",
    "BCC energy difference",
    "BCC effective lattice constant",
    "BCC Fermi",
    "BCC magnetic moment",
    "BCC volume per atom",
    "Heat capacity mass",
    "Heat capacity molar",
    "GS band gap",
    "GS estimated BCC lattice constant",
    "GS estimated FCC lattice constant",
    "GS magnetic moment",
    "GS volume per atom",
    "Is metal",
    "Is non-metal",
    "Is metalloid",
    "Is alkali",
    "Is d-block",
    "Is f-block",
    "No. of valence",
    "No. of unfilled",
    "No. of s valence",
    "No. of d unfilled",
    "No. of d valence",
    "No. of f unfilled",
    "No. of f valence",
    "No. of p valence",
    "No. of s unfilled",
    "No. of p unfilled"
]

# Categorical features
categorical_cols = [
    "Working_Ion",
    "Crystal_Lattice_Type"
]

# Numerical input features
numeric_cols = feature_cols + ["Ion_Concentration"]

# X = ALL INPUTS
X = df[numeric_cols + categorical_cols]

# y = TARGET
y = df["Voltage_V"]

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=X["Working_Ion"]
)

# Preprocessing
preprocessor = ColumnTransformer([
    (
        "numeric",
        StandardScaler(),
        numeric_cols
    ),
    (
        "categorical",
        OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        ),
        categorical_cols
    )
])

# FIT ONLY ON TRAINING DATA
X_train_processed = preprocessor.fit_transform(X_train)

# TRANSFORM TEST DATA
X_test_processed = preprocessor.transform(X_test)

print("Original dataset:", df.shape)
print("X before preprocessing:", X.shape)
print("X_train:", X_train_processed.shape)
print("X_test:", X_test_processed.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout
from tensorflow.keras.optimizers import Adam

dnn_model = Sequential([
    Input(shape=(X_train_processed.shape[1],)),

    Dense(128, activation="relu"),
    Dense(64, activation="relu"),
    Dense(32, activation="relu"),

    Dense(1)
])

dnn_model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss="mse",
    metrics=["mae"]
)

dnn_model.summary()
```

```python
history = dnn_model.fit(
    X_train_processed,
    y_train.to_numpy(),
    validation_split=0.20,
    epochs=100,
    batch_size=32,
    verbose=1
)
```

```python
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

dnn_mae = mean_absolute_error(y_test, y_pred_dnn)

dnn_rmse = np.sqrt(
    mean_squared_error(y_test, y_pred_dnn)
)

dnn_r2 = r2_score(y_test, y_pred_dnn)

print("========== DNN RESULTS ==========")
print(f"MAE  : {dnn_mae:.4f} V")
print(f"RMSE : {dnn_rmse:.4f} V")
print(f"R²   : {dnn_r2:.4f}")
```

```python
from sklearn.decomposition import PCA
import numpy as np
import matplotlib.pyplot as plt

pca_full = PCA()

# Fit ONLY on training data
pca_full.fit(X_train_processed)

explained_variance = pca_full.explained_variance_ratio_
cumulative_variance = np.cumsum(explained_variance)

print("Original number of features:", X_train_processed.shape[1])
print("Maximum possible PCs:", len(explained_variance))
```

```python
for threshold in [0.90, 0.95, 0.99]:
    n_components = np.argmax(cumulative_variance >= threshold) + 1
    
    print(
        f"{threshold*100:.0f}% explained variance → "
        f"{n_components} PCs"
    )
```

```python
#18 PC DATASET
from sklearn.decomposition import PCA

# PCA retaining 95% of the variance
pca = PCA(n_components=18)

# Fit ONLY on training data
X_train_pca = pca.fit_transform(X_train_processed)

# Transform test data using the same PCA
X_test_pca = pca.transform(X_test_processed)

print("Original training shape:", X_train_processed.shape)
print("PCA training shape:", X_train_pca.shape)

print("Original testing shape:", X_test_processed.shape)
print("PCA testing shape:", X_test_pca.shape)

print(
    "Explained variance by 18 PCs:",
    pca.explained_variance_ratio_.sum()
)
```

```python
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Input, Dense
from tensorflow.keras.optimizers import Adam

# DNN using PCA features
dnn_pca = Sequential([
    Input(shape=(18,)),
    Dense(128, activation="relu"),
    Dense(64, activation="relu"),
    Dense(32, activation="relu"),
    Dense(1)
])

dnn_pca.compile(
    optimizer=Adam(learning_rate=0.001),
    loss="mse",
    metrics=["mae"]
)

dnn_pca.summary()
```

```python
history_pca = dnn_pca.fit(
    X_train_pca,
    y_train,
    validation_split=0.20,
    epochs=100,
    batch_size=32,
    verbose=1
)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

y_pred_dnn_pca = dnn_pca.predict(X_test_pca).flatten()

mae_dnn_pca = mean_absolute_error(y_test, y_pred_dnn_pca)
rmse_dnn_pca = np.sqrt(mean_squared_error(y_test, y_pred_dnn_pca))
r2_dnn_pca = r2_score(y_test, y_pred_dnn_pca)

print("PCA-DNN Results")
print("----------------")
print(f"MAE  : {mae_dnn_pca:.6f} V")
print(f"RMSE : {rmse_dnn_pca:.6f} V")
print(f"R²   : {r2_dnn_pca:.6f}")
```

```python
feature_columns = [
    "Atomic volume",
    "Atomic number",
    "Atomic weight",
    "Covalent radius",
    "Oxidation states",
    "Boiling temperature",
    "Column number",
    "Row number",
    "First ionization energy",
    "Space group number",
    "Density",
    "Mendeleev number",
    "ICSD volume",
    "Polarisability",
    "BCC energy difference",
    "BCC effective lattice constant",
    "BCC Fermi",
    "BCC magnetic moment",
    "BCC volume per atom",
    "Heat capacity mass",
    "Heat capacity molar",
    "GS band gap",
    "GS estimated BCC lattice constant",
    "GS estimated FCC lattice constant",
    "GS magnetic moment",
    "GS volume per atom",
    "Is metal",
    "Is non-metal",
    "Is metalloid",
    "Is alkali",
    "Is d-block",
    "Is f-block",
    "No. of valence",
    "No. of unfilled",
    "No. of s valence",
    "No. of d unfilled",
    "No. of d valence",
    "No. of f unfilled",
    "No. of f valence",
    "No. of p valence",
    "No. of s unfilled",
    "No. of p unfilled"
]

input_columns = feature_columns + [
    "Ion_Concentration",
    "Working_Ion",
    "Crystal_Lattice_Type"
]

X_final = df[input_columns]
y_final = df["Voltage_V"]
```

```python
from sklearn.model_selection import train_test_split

X_train_90, X_test_10, y_train_90, y_test_10 = train_test_split(
    X_final,
    y_final,
    test_size=0.10,
    random_state=42,
    stratify=X_final["Working_Ion"]
)

print("Training data:", X_train_90.shape)
print("Testing data :", X_test_10.shape)
print("Training target:", y_train_90.shape)
print("Testing target :", y_test_10.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

numeric_features = feature_columns + ["Ion_Concentration"]

categorical_features = [
    "Working_Ion",
    "Crystal_Lattice_Type"
]

preprocessor_90 = ColumnTransformer(
    transformers=[
        (
            "num",
            StandardScaler(),
            numeric_features
        ),
        (
            "cat",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_features
        )
    ]
)
```

```python
X_train_90_processed = preprocessor_90.fit_transform(X_train_90)

X_test_10_processed = preprocessor_90.transform(X_test_10)

print("Processed training:", X_train_90_processed.shape)
print("Processed testing :", X_test_10_processed.shape)
```

```python
from sklearn.model_selection import KFold, GridSearchCV
from sklearn.svm import SVR
```

```python
cv = KFold(
    n_splits=10,
    shuffle=True,
    random_state=42
)
```

```python
svr = SVR(kernel="rbf")
```

```python
param_grid_svr = {
    "C": [1, 10, 100, 300],
    "gamma": ["scale", 0.01, 0.001],
    "epsilon": [0.05, 0.1, 0.2]
}
```

```python
grid_svr = GridSearchCV(
    estimator=svr,
    param_grid=param_grid_svr,
    scoring="neg_mean_absolute_error",
    cv=cv,
    n_jobs=-1,
    verbose=1
)

grid_svr.fit(X_train_90_processed, y_train_90)
```

```python
print("Best parameters:")
print(grid_svr.best_params_)

print("\nBest CV MAE:")
print(-grid_svr.best_score_)
```
