```python
# Print the Magpie properties in a clean numbered list

for i, prop in enumerate(magpie_properties, start=1):
    print(f"{i:3}. {prop}")
```

```python
# Search Magpie properties for anything related to BCC, lattice, or Fermi

keywords = ["BCC", "bcc", "lattice", "Fermi", "fermi"]

for prop in magpie_properties:
    if any(keyword in prop for keyword in keywords):
        print(prop)
```

```python
import pandas as pd
import requests
from io import StringIO

url = "https://raw.githubusercontent.com/nissy-dev/machine-learning-the-voltage-of-electrode-materials/master/dataset/mp_data/data_2019_12_03.csv"

response = requests.get(url)

print("Status:", response.status_code)

mp_df = pd.read_csv(StringIO(response.text))

print("Shape:", mp_df.shape)
print("\nColumns:")
for i, col in enumerate(mp_df.columns, start=1):
    print(i, "→", col)
```

```python
import os
import requests

base_url = "https://raw.githubusercontent.com/nissy-dev/machine-learning-the-voltage-of-electrode-materials/master/dataset/element_data/"

element_files = [
    "AtimicNumber.table",
    "AtomicVolume.table",
    "AtomicWeight.table",
    "BCCefflatcnt.table",
    "BCCenergydiff.table",
    "BCCfermi.table",
    "BCCmagmom.table",
    "BCCvolume_pa.table",
    "BCCvolume_padiff.table",
    "BoilingT.table",
    "ColumnNumber.table",
    "CovalentRadius.table",
    "Density.table",
    "FirstIonizationEnergy.table",
    "GSbandgap.table",
    "GSefflatcnt.table",
    "GSestBCClatcnt.table",
    "GSestFCClatcnt.table",
    "GSmagmom.table",
    "GSvolume_pa.table",
    "HeatCapacityMass.table",
    "HeatCapacityMolar.table",
    "ICSDVolume.table",
    "IsAlkali.table",
    "IsDBlock.table",
    "IsFBlock.table",
    "IsMetal.table",
    "IsMetalloid.table",
    "IsNonmetal.table",
    "MendeleevNumber.table",
    "NUnfilled.table",
    "NValance.table",
    "NdUnfilled.table",
    "NdValence.table",
    "NfUnfilled.table",
    "NfValence.table",
    "NpUnfilled.table",
    "NpValence.table",
    "NsUnfilled.table",
    "NsValence.table",
    "OxidationStates.table",
    "Polarizability.table",
    "RowNumber.table",
    "SpaceGroupNumber.table"
]

os.makedirs("element_data", exist_ok=True)

for filename in element_files:
    url = base_url + filename
    response = requests.get(url)

    if response.status_code == 200:
        with open(os.path.join("element_data", filename), "wb") as f:
            f.write(response.content)
        print("Downloaded:", filename)
    else:
        print("FAILED:", filename, response.status_code)

print("\nFinished.")
```

```python
import h5py

h5_path = "data_2019_12_03.h5"

with h5py.File(h5_path, "r") as f:
    h5_ids = set(f.keys())

charge_ids = set(mp_df["id_charge"])
discharge_ids = set(mp_df["id_discharge"])

missing_charge = charge_ids - h5_ids
missing_discharge = discharge_ids - h5_ids

print("Charge IDs:", len(charge_ids))
print("Discharge IDs:", len(discharge_ids))

print("Missing charge structures:", len(missing_charge))
print("Missing discharge structures:", len(missing_discharge))

if missing_charge:
    print("Example missing charge IDs:", list(missing_charge)[:10])

if missing_discharge:
    print("Example missing discharge IDs:", list(missing_discharge)[:10])
```

```python
import os
import requests

os.makedirs("preprocess", exist_ok=True)

files = {
    "create_features.py":
        "https://raw.githubusercontent.com/nissy-dev/machine-learning-the-voltage-of-electrode-materials/master/preprocess/create_features.py",

    "utils.py":
        "https://raw.githubusercontent.com/nissy-dev/machine-learning-the-voltage-of-electrode-materials/master/preprocess/utils.py"
}

for filename, url in files.items():
    response = requests.get(url)
    response.raise_for_status()

    with open(os.path.join("preprocess", filename), "wb") as f:
        f.write(response.content)

    print("Downloaded:", filename)

print("\nPreprocess folder created successfully.")
```
