```python
# ============================================================
# ML-MiB OFFLINE DATASET GENERATOR
# 239 FEATURES + average_voltage
# ============================================================

import os
import h5py
import numpy as np
import pandas as pd

from pymatgen.core import Composition, Element, Structure
from pymatgen.symmetry.analyzer import SpacegroupAnalyzer


# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

PROJECT = r"C:\Users\ADMIN\Desktop\PS PROJECT\PS 2-1 CRYSTAL DOMAIN\PROJECT"
PROGRAM = r"C:\Users\ADMIN\Desktop\PS PROJECT\PS 2-1 CRYSTAL DOMAIN\PROJECT\PROGRAM"

CSV_PATH = os.path.join(
    PROJECT,
    PROGRAM,
    "dataset",
    "mp_data",
    "data_2019_12_03.csv"
)

H5_PATH = os.path.join(
    PROJECT,
    "data_2019_12_03.h5"
)

ELEMENT_PATH = os.path.join(
    PROJECT,
    "element_data"
)

OUTPUT_PATH = os.path.join(
    PROJECT,
    "ML_MiB_3977_239_features.csv"
)


# ------------------------------------------------------------
# 2. 44 ORIGINAL ELEMENT FEATURES
# ------------------------------------------------------------

ELEMENT_FEATURES = [
    "AtimicNumber",
    "AtomicVolume",
    "AtomicWeight",
    "BCCefflatcnt",
    "BCCenergydiff",
    "BCCfermi",
    "BCCmagmom",
    "BCCvolume_pa",
    "BCCvolume_padiff",
    "BoilingT",
    "ColumnNumber",
    "CovalentRadius",
    "Density",
    "FirstIonizationEnergy",
    "GSbandgap",
    "GSefflatcnt",
    "GSestBCClatcnt",
    "GSestFCClatcnt",
    "GSmagmom",
    "GSvolume_pa",
    "HeatCapacityMass",
    "HeatCapacityMolar",
    "ICSDVolume",
    "IsAlkali",
    "IsDBlock",
    "IsFBlock",
    "IsMetal",
    "IsMetalloid",
    "IsNonmetal",
    "MendeleevNumber",
    "NUnfilled",
    "NValance",
    "NdUnfilled",
    "NdValence",
    "NfUnfilled",
    "NfValence",
    "NpUnfilled",
    "NpValence",
    "NsUnfilled",
    "NsValence",
    "OxidationStates",
    "Polarizability",
    "RowNumber",
    "SpaceGroupNumber"
]


# ------------------------------------------------------------
# 3. LOAD ELEMENT TABLES
# ------------------------------------------------------------

element_tables = {}

for feature in ELEMENT_FEATURES:

    file_path = os.path.join(
        ELEMENT_PATH,
        feature + ".table"
    )

    if not os.path.exists(file_path):
        raise FileNotFoundError(
            f"Missing element table:\n{file_path}"
        )

    table = {}

    with open(file_path, "r", encoding="utf-8") as f:

        for line in f:

            line = line.strip()

            if not line:
                continue

            parts = line.split()

            if len(parts) < 2:
                continue

            symbol = parts[0]

            try:
                value = float(parts[1])

            except:
                value = 0.0

            table[symbol] = value

    element_tables[feature] = table


# ------------------------------------------------------------
# 4. ELEMENT FEATURE VECTOR
# ------------------------------------------------------------

def get_element_features(symbol):

    values = []

    for feature in ELEMENT_FEATURES:

        value = element_tables[feature].get(
            symbol,
            0.0
        )

        values.append(float(value))

    return np.array(values, dtype=float)


# ------------------------------------------------------------
# 5. FORMULA FEATURES
# ------------------------------------------------------------

def formula_features(formula):

    comp = Composition(formula)

    elements = comp.elements

    amounts = np.array(
        [comp[e] for e in elements],
        dtype=float
    )

    fractions = amounts / amounts.sum()

    matrix = np.array([
        get_element_features(str(e))
        for e in elements
    ])

    mean = np.sum(
        matrix * fractions[:, None],
        axis=0
    )

    deviation = np.sum(
        np.abs(matrix - mean) *
        fractions[:, None],
        axis=0
    )

    # Single element = 44
    if len(elements) == 1:
        return mean

    # Multiple elements = 88
    return np.concatenate([
        mean,
        deviation
    ])


# ------------------------------------------------------------
# 6. WORKING ION
# ------------------------------------------------------------

IONS = [
    "Li",
    "Ca",
    "Cs",
    "Rb",
    "K",
    "Y",
    "Na",
    "Al",
    "Zn",
    "Mg"
]


def ion_features(ion):

    x = np.zeros(10)

    if ion in IONS:
        x[IONS.index(ion)] = 1

    return x


# ------------------------------------------------------------
# 7. CRYSTAL SYSTEM
# ------------------------------------------------------------

CRYSTAL_SYSTEMS = [
    "triclinic",
    "monoclinic",
    "orthorhombic",
    "tetragonal",
    "trigonal",
    "hexagonal",
    "cubic"
]


def crystal_features(system):

    x = np.zeros(7)

    if system in CRYSTAL_SYSTEMS:
        x[CRYSTAL_SYSTEMS.index(system)] = 1

    return x


# ------------------------------------------------------------
# 8. READ H5 STRUCTURE
# ------------------------------------------------------------

def read_structure(h5, material_id):

    cif = h5[material_id][()]

    if isinstance(cif, bytes):
        cif = cif.decode("utf-8")

    return Structure.from_str(
        cif,
        fmt="cif"
    )


# ------------------------------------------------------------
# 9. GENERATE 239 FEATURES
# ------------------------------------------------------------

def make_features(
    charge_structure,
    discharge_structure,
    working_ion
):

    # Crystal information
    analyzer = SpacegroupAnalyzer(
        charge_structure,
        symprec=0.1
    )

    space_group = analyzer.get_space_group_number()

    crystal_system = analyzer.get_crystal_system()

    crystal = np.concatenate([
        np.array([space_group], dtype=float),
        crystal_features(crystal_system)
    ])

    # Working ion
    ion = ion_features(working_ion)

    # Ion concentration
    discharge_comp = Composition(
        discharge_structure.formula
    )

    concentration = np.array([
        discharge_comp.get_atomic_fraction(
            Element(working_ion)
        )
    ])

    # Charge
    charge = formula_features(
        charge_structure.formula
    )

    if len(charge) == 44:
        charge = np.concatenate([
            charge,
            np.zeros(44)
        ])

    # Discharge
    discharge = formula_features(
        discharge_structure.formula
    )

    if len(discharge) == 44:
        discharge = np.concatenate([
            discharge,
            np.zeros(44)
        ])

    # Working-ion elemental properties
    working_ion_properties = get_element_features(
        working_ion
    )

    # Combine
    x = np.concatenate([
        crystal,                  # 8
        ion,                      # 10
        concentration,            # 1
        charge,                   # 88
        discharge,                # 88
        working_ion_properties    # 44
    ])

    if len(x) != 239:
        raise ValueError(
            f"Expected 239 features, got {len(x)}"
        )

    return x


# ------------------------------------------------------------
# 10. LOAD RAW DATA
# ------------------------------------------------------------

df = pd.read_csv(CSV_PATH)

print("Raw CSV:", df.shape)


# ------------------------------------------------------------
# 11. REMOVE EXACT DUPLICATE INSTANCES
# ------------------------------------------------------------

df = df.drop_duplicates(
    subset=[
        "id_charge",
        "id_discharge",
        "working_ion",
        "average_voltage"
    ]
).reset_index(drop=True)

print(
    "After duplicate removal:",
    df.shape
)


# ------------------------------------------------------------
# 12. GENERATE FEATURES
# ------------------------------------------------------------

rows = []
failed = []

with h5py.File(H5_PATH, "r") as h5:

    total = len(df)

    for i, row in df.iterrows():

        try:

            charge_structure = read_structure(
                h5,
                row["id_charge"]
            )

            discharge_structure = read_structure(
                h5,
                row["id_discharge"]
            )

            x = make_features(
                charge_structure,
                discharge_structure,
                row["working_ion"]
            )

            rows.append(
                np.concatenate([
                    x,
                    [row["average_voltage"]]
                ])
            )

        except Exception as e:

            failed.append(
                (i, str(e))
            )

        if (i + 1) % 100 == 0:

            print(
                f"Processed {i+1}/{total}"
            )


# ------------------------------------------------------------
# 13. CREATE DATAFRAME
# ------------------------------------------------------------

columns = [
    f"feat_{i}"
    for i in range(1, 240)
]

columns.append("average_voltage")

final_df = pd.DataFrame(
    rows,
    columns=columns
)

print()
print("Generated dataset:", final_df.shape)
print("Failed:", len(failed))


# ------------------------------------------------------------
# 14. SAFETY CHECK
# ------------------------------------------------------------

if len(final_df) != 3977:

    raise RuntimeError(
        "\nSTOPPED SAFELY!\n"
        f"The generated dataset has {len(final_df)} rows, "
        "not 3977.\n\n"
        "I will NOT randomly delete rows to force it to 3977.\n"
        "The exact paper cleaning rule must be identified first."
    )


if len(final_df.columns) != 240:

    raise RuntimeError(
        f"Expected 240 columns, got "
        f"{len(final_df.columns)}"
    )


# ------------------------------------------------------------
# 15. SAVE FINAL CSV
# ------------------------------------------------------------

final_df.to_csv(
    OUTPUT_PATH,
    index=False
)

print()
print("==========================================")
print("FINAL CSV CREATED SUCCESSFULLY")
print("==========================================")
print("Rows   :", len(final_df))
print("Columns:", len(final_df.columns))
print("Features: 239")
print("Target : average_voltage")
print()
print("Saved at:")
print(OUTPUT_PATH)
```
