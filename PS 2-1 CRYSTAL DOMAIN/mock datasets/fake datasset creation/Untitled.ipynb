```python
import requests
import os

# GitHub raw CSV URL
url = "https://raw.githubusercontent.com/nissy-dev/machine-learning-the-voltage-of-electrode-materials/master/dataset/mp_data/data_2019_12_03.csv"

# Folder where the CSV will be saved
folder = "../dataset/mp_data"
os.makedirs(folder, exist_ok=True)

# Output file
output_path = os.path.join(folder, "data_2019_12_03.csv")

print("Downloading ML-MiB dataset...")

response = requests.get(url, stream=True)
response.raise_for_status()

total_size = int(response.headers.get("content-length", 0))
downloaded = 0

with open(output_path, "wb") as f:
    for chunk in response.iter_content(chunk_size=8192):
        if chunk:
            f.write(chunk)
            downloaded += len(chunk)

            if total_size:
                percent = downloaded / total_size * 100
                print(f"\rDownloading: {percent:.2f}%", end="")

print("\n\n✅ Download completed!")
print("Saved at:", os.path.abspath(output_path))
```

```python
import os
import sys
import h5py
import numpy as np
import pandas as pd

from tqdm.auto import tqdm

from pymatgen.core import Composition, Element, Structure
from pymatgen.symmetry.analyzer import SpacegroupAnalyzer

# ---------------------------------------------------------
# 1. PATHS
# ---------------------------------------------------------

PROGRAM = r"C:\Users\ADMIN\Desktop\PS PROJECT\PS 2-1 CRYSTAL DOMAIN\PROJECT\PROGRAM"

# Raw battery CSV
BATTERY_CSV = os.path.join(
    PROGRAM,
    "dataset",
    "mp_data",
    "data_2019_12_03.csv"
)

# Your H5 file
H5_FILE = os.path.join(
    PROGRAM,
    "data_2019_12_03.h5"
)

# 44 elemental-property tables
ELEMENT_DATA = os.path.join(
    PROGRAM,
    "element_data"
)

# Output CSV
OUTPUT_CSV = os.path.join(
    PROGRAM,
    "dataset",
    "mp_data",
    "ML_MiB_239_features.csv"
)

# ---------------------------------------------------------
# 2. IMPORT REPOSITORY UTILS.PY
# ---------------------------------------------------------

PREPROCESS_FOLDER = os.path.join(PROGRAM, "preprocess")

if PREPROCESS_FOLDER not in sys.path:
    sys.path.insert(0, PREPROCESS_FOLDER)

from utils import (
    NUM_OF_FEATURE,
    ElementDataSingleton,
    create_feature_from_formula,
    create_feature_from_wokring_ion,
    create_feature_from_crystal_system
)

print("Repository utilities loaded.")
print("NUM_OF_FEATURE =", NUM_OF_FEATURE)

# ---------------------------------------------------------
# 3. CHECK FILES
# ---------------------------------------------------------

print("\nChecking files...")

required_files = {
    "Battery CSV": BATTERY_CSV,
    "H5 file": H5_FILE,
    "Element data folder": ELEMENT_DATA
}

for name, file_path in required_files.items():

    if os.path.exists(file_path):
        print(f"✅ {name}: FOUND")
    else:
        print(f"❌ {name}: NOT FOUND")
        print(file_path)

# ---------------------------------------------------------
# 4. LOAD BATTERY DATA
# ---------------------------------------------------------

print("\nLoading battery CSV...")

battery_table = pd.read_csv(
    BATTERY_CSV,
    index_col=False
)

print("Rows:", len(battery_table))
print("Columns:", len(battery_table.columns))

print("\nColumns:")
print(battery_table.columns.tolist())

# ---------------------------------------------------------
# 5. LOAD ELEMENT DATA
# ---------------------------------------------------------

print("\nLoading elemental-property tables...")

element_data = ElementDataSingleton.get_instance(
    element_data_path=ELEMENT_DATA
)

print("✅ Element data loaded.")

# ---------------------------------------------------------
# 6. OPEN H5
# ---------------------------------------------------------

print("\nOpening H5 file...")

cif_data = h5py.File(
    H5_FILE,
    "r"
)

print("Number of structures in H5:", len(cif_data))

# ---------------------------------------------------------
# 7. PREPARE FEATURE ARRAY
# ---------------------------------------------------------

n_rows = len(battery_table)

features = np.zeros(
    (n_rows, NUM_OF_FEATURE),
    dtype=float
)

print("\nFeature array created:")
print(features.shape)

# ---------------------------------------------------------
# 8. CACHE
# ---------------------------------------------------------

# Cache crystal information because the same material ID
# can appear multiple times.

spacegroup_cache = {}
crystal_system_cache = {}

# Cache composition features
formula_cache = {}

# ---------------------------------------------------------
# 9. HELPER FUNCTION
# ---------------------------------------------------------

def get_cif_structure(material_id):

    cif = cif_data[material_id][()]

    if isinstance(cif, bytes):
        cif = cif.decode("utf-8")

    return Structure.from_str(
        cif,
        fmt="cif"
    )


def get_formula_features(formula):

    if formula not in formula_cache:

        formula_cache[formula] = create_feature_from_formula(
            formula,
            element_data
        )

    return formula_cache[formula].copy()


# ---------------------------------------------------------
# 10. GENERATE FEATURES
# ---------------------------------------------------------

print("\n==============================================")
print("STARTING ML-MiB FEATURE GENERATION")
print("==============================================")

progress = tqdm(
    total=n_rows,
    desc="Generating features",
    unit="row",
    dynamic_ncols=True
)

start_time = __import__("time").time()

try:

    for i in range(n_rows):

        # -------------------------------------------------
        # Current row
        # -------------------------------------------------

        row = battery_table.iloc[i]

        id_charge = row["id_charge"]
        id_discharge = row["id_discharge"]
        ion = row["working_ion"]

        progress.set_description(
            f"Processing {i + 1}/{n_rows}"
        )

        # -------------------------------------------------
        # CHARGE STRUCTURE
        # -------------------------------------------------

        if id_charge in spacegroup_cache:

            space_group_number = spacegroup_cache[id_charge]
            crystal_system = crystal_system_cache[id_charge]

        else:

            charge_crystal = get_cif_structure(
                id_charge
            )

            # symprec=0.1 helps make the calculation
            # considerably more practical on modern pymatgen

            finder = SpacegroupAnalyzer(
                charge_crystal,
                symprec=0.1
            )

            space_group_number = (
                finder.get_space_group_number()
            )

            crystal_system = (
                finder.get_crystal_system()
            )

            spacegroup_cache[id_charge] = (
                space_group_number
            )

            crystal_system_cache[id_charge] = (
                crystal_system
            )

        # -------------------------------------------------
        # SPACE GROUP FEATURE
        # -------------------------------------------------

        space_group_feat = np.array([
            space_group_number
        ])

        # -------------------------------------------------
        # CRYSTAL SYSTEM FEATURE
        # -------------------------------------------------

        crystal_system_feat = (
            create_feature_from_crystal_system(
                crystal_system
            )
        )

        # -------------------------------------------------
        # WORKING ION FEATURE
        # -------------------------------------------------

        ion_feat = (
            create_feature_from_wokring_ion(
                ion
            )
        )

        # -------------------------------------------------
        # DISCHARGE STRUCTURE
        # -------------------------------------------------

        discharge_crystal = get_cif_structure(
            id_discharge
        )

        discharge_composition = Composition(
            discharge_crystal.formula
        )

        ion_concentrate = np.array([
            discharge_composition.get_atomic_fraction(
                Element(ion)
            )
        ])

        # -------------------------------------------------
        # FORMULAS
        # -------------------------------------------------

        charge_formula = charge_crystal.formula \
            if "charge_crystal" in locals() else \
            get_cif_structure(id_charge).formula

        discharge_formula = discharge_crystal.formula

        # -------------------------------------------------
        # CHARGE COMPOSITION FEATURES
        # -------------------------------------------------

        feature_charge = get_formula_features(
            charge_formula
        )

        # If only one element -> pad 44 zeros
        if feature_charge.shape[0] == 44:

            feature_charge = np.concatenate([
                feature_charge,
                np.zeros(44)
            ])

        # -------------------------------------------------
        # DISCHARGE COMPOSITION FEATURES
        # -------------------------------------------------

        feature_discharge = get_formula_features(
            discharge_formula
        )

        # Make sure it has 88 features
        if feature_discharge.shape[0] == 44:

            feature_discharge = np.concatenate([
                feature_discharge,
                np.zeros(44)
            ])

        # -------------------------------------------------
        # WORKING ION ELEMENT FEATURES
        # -------------------------------------------------

        feature_ion = get_formula_features(
            ion
        )

        # -------------------------------------------------
        # COMBINE ALL FEATURES
        # -------------------------------------------------

        complete_feature = np.concatenate([
            space_group_feat,
            crystal_system_feat,
            ion_feat,
            ion_concentrate,
            feature_charge,
            feature_discharge,
            feature_ion
        ])

        # -------------------------------------------------
        # CHECK FEATURE COUNT
        # -------------------------------------------------

        if len(complete_feature) != NUM_OF_FEATURE:

            raise ValueError(
                f"Feature count error at row {i}. "
                f"Expected {NUM_OF_FEATURE}, "
                f"got {len(complete_feature)}"
            )

        # -------------------------------------------------
        # SAVE INTO ARRAY
        # -------------------------------------------------

        features[i] = complete_feature

        # -------------------------------------------------
        # UPDATE PROGRESS
        # -------------------------------------------------

        progress.update(1)

finally:

    progress.close()

# ---------------------------------------------------------
# 11. CREATE FEATURE DATAFRAME
# ---------------------------------------------------------

print("\nCreating DataFrame...")

feature_columns = [
    f"feat_{i + 1}"
    for i in range(NUM_OF_FEATURE)
]

feature_df = pd.DataFrame(
    features,
    columns=feature_columns
)

# ---------------------------------------------------------
# 12. ADD TARGET VOLTAGE
# ---------------------------------------------------------

feature_df["average_voltage"] = (
    battery_table["average_voltage"].values
)

# ---------------------------------------------------------
# 13. SAVE CSV
# ---------------------------------------------------------

print("\nSaving CSV...")

feature_df.to_csv(
    OUTPUT_CSV,
    index=False
)

# ---------------------------------------------------------
# 14. FINAL INFORMATION
# ---------------------------------------------------------

elapsed = __import__("time").time() - start_time

print("\n==============================================")
print("✅ ML-MiB DATASET CREATED")
print("==============================================")

print("Rows:", feature_df.shape[0])
print("Columns:", feature_df.shape[1])

print("Feature columns:", NUM_OF_FEATURE)
print("Target column: average_voltage")

print(
    f"Time taken: {elapsed / 60:.2f} minutes"
)

print("\nSaved file:")
print(OUTPUT_CSV)

print("\nFirst 5 rows:")
display(feature_df.head())
```

```python
import os
import pandas as pd
from tqdm.auto import tqdm

# ============================================================
# 1. PROGRAM PATH
# ============================================================

PROGRAM = r"C:\Users\ADMIN\Desktop\PS PROJECT\PS 2-1 CRYSTAL DOMAIN\PROJECT\PROGRAM"

# Your folder containing the 44 .table files
ELEMENT_DATA = os.path.join(
    PROGRAM,
    "element_data"
)

# Output CSV
OUTPUT_CSV = os.path.join(
    PROGRAM,
    "element_properties_44.csv"
)

# ============================================================
# 2. EXACT 44 FEATURE NAMES
# ============================================================

FEATURE_NAMES = [
    "AtimicNumber",
    "AtomicVolume",
    "AtomicWeight",
    "BCCefflatcnt",
    "BCCenergydiff",
    "BCCfermi",
    "BCCmagmom",
    "BCCvolume_pa",
    "BCCvolume_padiff",
    "BoilingT",
    "ColumnNumber",
    "CovalentRadius",
    "Density",
    "FirstIonizationEnergy",
    "GSbandgap",
    "GSefflatcnt",
    "GSestBCClatcnt",
    "GSestFCClatcnt",
    "GSmagmom",
    "GSvolume_pa",
    "HeatCapacityMass",
    "HeatCapacityMolar",
    "ICSDVolume",
    "IsAlkali",
    "IsDBlock",
    "IsFBlock",
    "IsMetal",
    "IsMetalloid",
    "IsNonmetal",
    "MendeleevNumber",
    "NUnfilled",
    "NValance",
    "NdUnfilled",
    "NdValence",
    "NfUnfilled",
    "NfValence",
    "NpUnfilled",
    "NpValence",
    "NsUnfilled",
    "NsValence",
    "OxidationStates",
    "Polarizability",
    "RowNumber",
    "SpaceGroupNumber"
]

# ============================================================
# 3. CHECK ELEMENT DATA FOLDER
# ============================================================

print("Checking element_data folder...\n")

if not os.path.exists(ELEMENT_DATA):
    raise FileNotFoundError(
        f"❌ element_data folder not found:\n{ELEMENT_DATA}"
    )

print("✅ element_data folder found:")
print(ELEMENT_DATA)

# ============================================================
# 4. CHECK ALL 44 FILES
# ============================================================

print("\nChecking 44 feature files...\n")

missing_files = []

for feature in FEATURE_NAMES:

    file_path = os.path.join(
        ELEMENT_DATA,
        feature + ".table"
    )

    if not os.path.exists(file_path):
        missing_files.append(feature + ".table")

if missing_files:

    print("❌ Missing files:\n")

    for file in missing_files:
        print("   ", file)

    raise FileNotFoundError(
        "\nSome required .table files are missing."
    )

print("✅ All 44 feature files found!")

# ============================================================
# 5. FUNCTION TO READ A .TABLE FILE
# ============================================================

def read_table_file(file_path):

    values = {}

    with open(
        file_path,
        "r",
        encoding="utf-8",
        errors="ignore"
    ) as file:

        for line in file:

            line = line.strip()

            # Skip empty lines
            if not line:
                continue

            # Skip comments
            if line.startswith("#"):
                continue

            parts = line.split()

            # Need at least element + value
            if len(parts) < 2:
                continue

            element = parts[0]

            # Everything after element
            raw_value = " ".join(parts[1:])

            # --------------------------------------------
            # Try converting to number
            # --------------------------------------------

            try:

                value = float(raw_value)

            except ValueError:

                # Some tables may contain non-numeric/
                # missing values.
                #
                # For ML numerical processing,
                # convert those to 0.

                value = 0.0

            values[element] = value

    return values

# ============================================================
# 6. LOAD ALL 44 FEATURES
# ============================================================

all_data = {}

print("\nLoading 44 elemental properties...\n")

for feature in tqdm(
    FEATURE_NAMES,
    desc="Loading features",
    unit="feature"
):

    file_path = os.path.join(
        ELEMENT_DATA,
        feature + ".table"
    )

    all_data[feature] = read_table_file(
        file_path
    )

print("\n✅ All 44 feature tables loaded!")

# ============================================================
# 7. FIND ALL ELEMENTS
# ============================================================

print("\nFinding elements...")

all_elements = set()

for feature in FEATURE_NAMES:

    all_elements.update(
        all_data[feature].keys()
    )

# Sort alphabetically
all_elements = sorted(all_elements)

print(
    f"✅ Found {len(all_elements)} elements."
)

# ============================================================
# 8. CREATE DATAFRAME
# ============================================================

print("\nCreating DataFrame...")

data = {}

# Element identifier
data["Element"] = all_elements

# Add the 44 actual feature names
for feature in FEATURE_NAMES:

    data[feature] = [
        all_data[feature].get(
            element,
            0.0
        )
        for element in all_elements
    ]

df = pd.DataFrame(data)

# ============================================================
# 9. CHECK COLUMN NAMES
# ============================================================

print("\n==========================================")
print("COLUMN CHECK")
print("==========================================")

print(
    "Number of elemental features:",
    len(FEATURE_NAMES)
)

print(
    "Total CSV columns including Element:",
    len(df.columns)
)

print("\nColumn names:\n")

for i, column in enumerate(df.columns, start=1):

    print(
        f"{i:02d}. {column}"
    )

# ============================================================
# 10. SAVE CSV
# ============================================================

df.to_csv(
    OUTPUT_CSV,
    index=False
)

# ============================================================
# 11. FINAL RESULT
# ============================================================

print("\n==========================================")
print("✅ CSV CREATED SUCCESSFULLY")
print("==========================================")

print(
    "Rows:",
    len(df)
)

print(
    "44 feature columns:",
    len(FEATURE_NAMES)
)

print(
    "Total columns including Element:",
    len(df.columns)
)

print("\nSaved at:")

print(OUTPUT_CSV)

# ============================================================
# 12. DISPLAY FIRST 10 ROWS
# ============================================================

print("\nFirst 10 rows:")

display(df.head(10))
```
