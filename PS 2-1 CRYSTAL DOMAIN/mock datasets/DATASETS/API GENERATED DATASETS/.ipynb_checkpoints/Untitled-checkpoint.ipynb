```python
import pandas as pd

# 1. Download the cleaned battery dataset directly via URL
url = "https://raw.githubusercontent.com/tikam24/wolverton_ml/main/cleaned_data.csv"
df = pd.read_csv(url)

# 2. Display dataset information and the first 5 rows
print(f"Dataset successfully loaded with {df.shape[0]} rows and {df.shape[1]} columns.\n")
display(df.head())

# 3. Save as a local CSV file
df.to_csv("battery_dataset.csv", index=False)
print("Saved locally as 'battery_dataset.csv'.")
```

```python
import pandas as pd
from mp_api.client import MPRester

API_KEY = ""  # Get key from https://materialsproject.org/api

# Connect to Materials Project API
with MPRester(API_KEY) as mpr:
    # Query insertion battery materials
    battery_docs = mpr.insertion_electrodes.search()
    
    # Extract relevant fields into a structured dictionary list
    data = []
    for doc in battery_docs:
        data.append({
            "battery_id": getattr(doc, "battery_id", None),
            "battery_formula": getattr(doc, "battery_formula", None),
            "working_ion": getattr(doc, "working_ion", None),
            "max_voltage": getattr(doc, "max_voltage", None),
            "capacity_grav": getattr(doc, "capacity_grav", None),
            "capacity_vol": getattr(doc, "capacity_vol", None),
            "num_steps": getattr(doc, "num_steps", None),
        })

# Convert to DataFrame
df_mp = pd.DataFrame(data)

# Save to CSV and view output
df_mp.to_csv("mp_battery_dataset.csv", index=False)
print(f"Downloaded {len(df_mp)} entries and saved to 'mp_battery_dataset.csv'.")
display(df_mp.head())
```
